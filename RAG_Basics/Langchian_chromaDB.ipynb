```python
from langchain_core.documents import Document
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceBgeEmbeddings
```

```python
documents = [
    Document(page_content = "Python is used for AI."),
    Document(page_content="Langchain helps build AI apps"),
    Document(page_content = "ChromaDB stores embeddings")

]
print(documents)
```

```python
embedding= HuggingFaceBgeEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
db = Chroma.from_documents(
    documents,
    embedding
)
```

```python
results = db.similarity_search("What is Lanchain?")
print(results[0].page_content)
```

```python
results = db.similarity_search("AI")

print(results[0].page_content)
```

```python
results = db.similarity_search("vector database")

print(results[0].page_content)
```

```python
db.add_texts(
    ['FastAPI is used for backend APIs.']
)
# old doc -> new Doc -> updated vector db
```

```python
results = db.similarity_search("backend")

print(results[0].page_content)
```
