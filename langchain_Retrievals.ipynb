```python
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchian_community.vectorestores import Chroma
```

```python
# create documents
docs = [
    Document(page_content="Machine learning is a subset of AI"),

    Document(page_content="Python is widely used in AI"),

    Document(page_content="Neural networks are inspired by the brain")
]
```

```python
#load model

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
from langchain_community.vectorstores import Chroma
```

```python
#create vectore store

vectorstore = Chroma.from_documents(
    documents=docs,
    embedding=embedding_model,
    persist_directory="my_chroma_db"
)
```

```python
retriever = vectorstore.as_retriever() #create retriever
```

```python
results = retriever.invoke("What is AI?")  #do the searching and all bts
```

```python
for doc in results:
    print(doc.page_content)  #  printitng results
```

```python
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
docs = [

    Document(
        page_content="AI is transforming healthcare"
    ),

    Document(
        page_content="AI improves hospital systems"
    ),

    Document(
        page_content="AI helps doctors diagnose diseases"
    ),

    Document(
        page_content="Pizza is an Italian food"
    ),

    Document(
        page_content="Machine learning powers recommendation systems"
    )
]
```

```python
# load embedding
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
#vectore  store
vectorstore = FAISS.from_documents(
    docs,
    embedding_model
)
```

```python
#create mmr retrieval
retriever = vectorstore.as_retriever(

    search_type="mmr",

    search_kwargs={
        "k": 2,
        "fetch_k": 5,
        "lambda_mult":0.5
    }
)
```

```python
results = retriever.invoke(
    "How is AI used in healthcare?"
)
```

```python
for i, doc in enumerate(results):

    print("\n" + "="*50)
    print(f"RESULT {i+1}")
    print("="*50)

    print(doc.page_content)
```
