```python
import os

from youtube_transcript_api import (
    YouTubeTranscriptApi,
    TranscriptsDisabled
)

from langchain_text_splitters import RecursiveCharacterTextSplitter

from langchain_huggingface import HuggingFaceEmbeddings

from langchain_community.vectorstores import FAISS

from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
os.environ["GOOGLE_API_KEY"] = ""
```

#1.Indexing

```python
video_id = "UU1WVnMk4E8"

try:
    ytt_api = YouTubeTranscriptApi()

    transcript_list = ytt_api.fetch(video_id)

    transcript = " ".join(
        chunk.text for chunk in transcript_list
    )

    print(transcript[:1000])

except TranscriptsDisabled:
    print("No captions available for this video.")
```

```python
transcript_list
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap = 200
)
chunks = text_splitter.split_text(transcript)

print("Total Chunks:", len(chunks))

print("\nFIRST CHUNK:\n")
print(chunks[342])
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
vectorstore  = FAISS.from_texts(
    chunks,
    embeddings
)
```

```python
vectorstore
```

```python
for i, chunk in enumerate(chunks):

    print("\n" + "="*80)
    print(f"CHUNK {i+1}")
    print("="*80)

    print(chunk)
```

```python
print(chunks[1])
```

```python
all_vectors = embeddings.embed_documents(chunks)

chunk_id = 10

print("="*80)
print(f"CHUNK ID: {chunk_id}")
print("="*80)

print("\nCHUNK TEXT:\n")
print(chunks[chunk_id])

print("\nVECTOR:\n")
print(all_vectors[chunk_id])

print("\nVECTOR DIMENSION:")
print(len(all_vectors[chunk_id]))
```

2.Retrieval

```python
results = vectorstore.similarity_search(
    query = "what is LLM?",
    k=3
)
```

```python
for i,doc in enumerate(results):
  print("\n"+ "="*50)
  print(f"Results {i+1}")
  print("-"*50)

  print(doc.page_content[:1000])
```

```python
queries = [
    "What is tokenization?",
    "Explain embeddings",
    "What is batch size?",
    "What is CUDA?"
]

for query in queries:

    print("\n" + "="*80)
    print("QUERY:", query)
    print("="*80)

    results = vectorstore.similarity_search(
        query,
        k=2
    )

    for i, doc in enumerate(results):

        print("\n" + "-"*60)
        print(f"RESULT {i+1}")
        print("-"*60)

        print(doc.page_content[:500])
```

3.Augmentation

```python
query = "What is batching in language models?"
```

```python
results = vectorstore.similarity_search(
    query,
    k=3
)
```

```python
context = "\n\n".join(
    [doc.page_content for doc in results]
)
```

```python
print(context)
```

```python
prompt = f"""
Answer the question based only on the context below.

Context:
{context}

Question:
{query}
"""
```

```python
print(prompt)
```

4.Generation

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash"
)

response = llm.invoke(prompt)

print(response.content)
```

Chain

```python
from langchain_core.prompts import ChatPromptTemplate

from langchain_core.output_parsers import StrOutputParser

from langchain_core.runnables import RunnablePassthrough
```

```python
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 3}
)
```

```python
prompt = ChatPromptTemplate.from_template("""
Answer the question based only on the context below.

Context:
{context}

Question:
{question}
""")
```

```python
rag_chain = (
    {
        "context": retriever,
        "question": RunnablePassthrough()
    }
    | prompt
    | llm
    | StrOutputParser()
)
```

```python
response = rag_chain.invoke(
    "What is batching in language models?"
)

print(response)
```
